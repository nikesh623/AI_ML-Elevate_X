# OOP

##1. Introduction to NumPy

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr)
```

##2. Indexing

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[0])
print(arr[2])
print(arr[-1])
```

2D indexing

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(arr[0, 1])
print(arr[1, 2])
```

##3. Slicing

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[1:4])
```

##4. Data Type (dtype)

```python
import numpy as np

arr = np.array([10, 20, 30])

print(arr.dtype)
```

```python
arr = np.array([1.5, 2.5, 3.5])

print(arr.dtype)
```

```python
arr = np.array([1, 2, 3], dtype=float)

print(arr)
print(arr.dtype)
```

##5. Array Shape

```python
import numpy as np

arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr.shape)
```

##6. Reshape

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

new_arr = arr.reshape(2, 3)

print(new_arr)
```

```python
arr = np.array([1, 2, 3, 4, 5, 6])

print(arr.reshape(2, -1))
```

##7. Sorting in NumPy

```python
import numpy as np

arr = np.array([50, 10, 40, 20, 30])

sorted_arr = np.sort(arr)

print(sorted_arr)
```

##8. Array Search

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

result = np.where(arr > 25)

print(result)
```
