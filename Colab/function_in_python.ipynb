````python
## In the above example, the numbers of argument is fixed, thus while calling the function, if the no.of argument is not matched, then in this case
## you'll get an error

## To avoid this issue

## Variable length arugment
## positional and keywords arguments
````

````python
def print_numbers(*args):
    for number in args:
      print(number)
````

````python
print_numbers(1, 2, 3, 4, 5, 6, 'Elevate-X')
````

````python
def print_numbers(*rebanta):
    for number in rebanta:
      print(number)
````

````python
print_numbers("rebanta", "aryal")
````

In the above example, * means to collext positional arguments whereas rebanta is the variable name

# Lamda Function

Lamda Function in Python:

Lamda functions are small anonymous function defined using the keyword ```lamda``` . It can have any number of arguments but only one expression.
They are commonly used for short operations or as arguements to nigher order function

````python
from ast import Expression
## Syntax

lambda arguments: Expression
````

````python
def add(a, b):
    return a+b
````

````python
add(2, 3)
````

````python
add = lambda a, b: a+b
print(add(2, 3))
````
