Check Number of Dimensions in an Numpy Array?

```python
# Check Number of Dimensions in an Numpy Array?
import numpy as np

a = np.array(42)
b = np.array([1,2,3,4,5])
c = np.array([[1,2,3], [4,5,6]])
d = np.array([[[1,2,3], [4,5,6]], [[1,2,3], [4,5,6]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

```python
# Create an array of 5 dim and verify its dim

import numpy as np

arr = np.array([1,2,3,4], ndmin=64)

print(arr)
print('number of dimensions :', arr.ndim)
```
