Variables declaration

```python
# varaible_name = value
"""Variable Rules
1. Must start with a letter (A–Z, a–z) or underscore (_).
2. Cannot start with a number.
3. Can contain letters, numbers, and underscores.
4. Case-sensitive (Age and age are different).
5. Cannot use Python keywords (if, for, class, etc.)."""
a = 3
b = 4
c = a + b
print (c)
name = "Nikesh"
age = 21
salary = 50000.50
is_student = True
print(name)
print(age)
print(salary)
print(is_student)
```
